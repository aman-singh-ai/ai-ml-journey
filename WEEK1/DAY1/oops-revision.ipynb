
# Constructor : Initializes object [__init__()]
# self : current object

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
student1 = Student("Aman", 7)
student2 = Student("Satyam", 8)
print(student1.name)
print(student1.roll_number)
print(student2.name)
print(student2.roll_number)
```
