## Class: Blueprint or template for creating objects.
## Object : An object is an instance of a class.
## Address are different because both are different objects.

```python
class Student:
    pass
student1=Student()
student2=Student()
print(student1)
print(student2)
```


## Constructor : Initializes object [__init__()]
## self : current object

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
student = Student("Aman", 7)
print(student.name)
print(student.roll_number)
```

## instance variable : individual data

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
student1 = Student("Aman", 20)
student2 = Student("Satyam", 21)
print(student1.name, student1.age)
print(student2.name, student2.age)        
```
