
## Constructor : Initializes object __init__()
## self : current object

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number
student = Student("Aman", 7)
print(student.name)
print(student.roll_number)
```

## instance variable : individual data / common to objects.

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
student1 = Student("Aman", 20)
student2 = Student("Satyam", 21)
print(student1.name, student1.age)
print(student2.name, student2.age)        
```

# class variable : shared data / specific to each object.

```python
class Student:
    college_name = "Babu Banarasi Das University"

    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number

student1 = Student("Aman", 7)
student2 = Student("Satyam", 8)

print(student1.name, student1.roll_number, student1.college_name)
print(student2.name, student2.roll_number, student2.college_name)
```


# Method : is a function defined inside a class that performs an operation related to the object or class.
>In the given example study() is a method. 

```python
class Student:
    def study(self):
        print("Student is studying.")
```

# 1. Instance Method : is a method that operates on the data of a specific object and takes self as its first parameter.

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_number)
student = Student("Aman Singh", 7)
student.display()        
```

# 2.Class Method : A class method is a method that operates on the class itself and takes cls as its first parameter.

>self vs cls:
self → current object
cls  → current class

```python
class Student:
    def __init__(self, name, roll_number):
        self.name = name
        self.roll_number = roll_number

    def display(self):
        print("Name:", self.name)
        print("Roll Number:", self.roll_number)
student = Student("Aman Singh", 7)
student.display()        
```

```python
class Student:

    college = "ABC Institute"

    @classmethod
    def show_college(cls):
        print("College:", cls.college)


Student.show_college()
```

# 3.Static Method:A static method is a method that does not depend on the instance or class state and does not require self or cls.

>In python it's defined by @staticmethod

```python
class Student:
    @staticmethod
    def is_adult(age):
        return age >= 18
print(Student.is_adult(20))  # Output: True
print(Student.is_adult(15))  # Output: False
```

# Instance , Class & Static method.

```python
class Student:

    college = "BBDU"

    def __init__(self, name):
        self.name = name

    # 1.Instance Method--------------->
    def show_name(self):
        print("Name:", self.name)

    # 2.Class Method--------------------->
    @classmethod
    def show_college(cls):
        print("College:", cls.college)

    # 3.Static Method---------->
    @staticmethod
    def is_adult(age):
        return age >= 18

student1 = Student("Aman")
student1.show_name()
Student.show_college()
print(Student.is_adult(19))
```

# Inheritance is an OOP mechanism in which a child class inherits properties and methods from a parent class.

### When to Use Inheritance??
>Inheritance is useful when there is an "IS-A" relationship between two classes.
>>If you can logically form the sentence "X is a Y", then inheritance is appropriate.
              >>>Examples:Student IS-A Person ✅
                       >>>> Dog IS-A Animal ✅

### Main Benefits of Inheritance:
 >Code Reusability: Derived classes can automatically reuse code from the base class.
 >>Reduced Redundancy: Eliminates duplicate code across multiple classes.
 >>>Easy Maintainability: Modifying common functionality in the parent class instantly updates all child classes.
 >>>>Extensibility: Child classes can easily extend parent features by adding unique functionality.

```python
class person:
    def __init__(self, name, age):
        self.name = name
        self.age = age                                 

    def introduce(self):
        print("My name is", self.name)

class Student(person):    #It means student class is inheriting the person class so,
    pass                                            #Person = Parent/Base/Super class
                                                    #Student = Child/Derived/Sub class
student = Student("Aman", 20)

print(student.name)  # Output: Aman
print(student.age)   # Output: 20   
student.introduce()  # Output: My name is Aman
```
